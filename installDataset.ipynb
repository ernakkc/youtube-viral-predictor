### Download the latest version

```python
path = kagglehub.dataset_download("vedauv/what-makes-a-yt-video-go-viral-12000-vids")
print(f"Dataset downloaded to: {path}")
```

### Move the dataset to the project location

```python
import os
import shutil

if "./data/dataset.csv" not in os.listdir("./data"):
    os.makedirs("./data", exist_ok=True)
    try: 
        shutil.copy(f"{path}/viral_videos.csv", "./data/dataset.csv")
    except Exception as e:
        print(f"Dataset file not found in the downloaded path. Please check the downloaded files. Error: {e}")


```
